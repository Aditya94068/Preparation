```python
#Using melt keyward
# df.melt(id_vars=None, value_vars=None, var_name=None, value_name="value", col_level=None)
# df2 = df.melt(id_vars="Name",value_vars=["Math","Science","English"],var_name="Subject",value_name="Score")
# df2 = df.melt(id_vars = "Name",value_vars=["Math","Science","English"],var_name = "Subject",value_name = "Score")
df2 = df.melt(id_vars = "Name",value_vars = ["Math","Science","English"],var_name = "Subject",value_name = "Marks")
df2 = df.melt(id_vars = "Name",value_vars = ["Math","Science","English"],var_name = "Subject",value_name = "Marlks")
```
