```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85, 90, 78, 85, 92, 88, 75, 80],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
        "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
         "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})  

# df = pd.DataFrame({
#     "Department" :["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales",

```

```python
# df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].mean()
```

```python
# df.groupby("Team")["Salary"].sum() # Average per team
df.groupby("Department")["Salary"].sum()
```

```python
# df.groupby("Team")["Salary"].count()    # How many entries
df.groupby("Department")["Salary"].count()
```

```python
# df.groupby("Team")["Salary"].min()
df.groupby("Department")["Salary"].min()
```

```python
# df.groupby("Team")["Salary"].max()
df.groupby("Department")["Salary"].max()
```

```python
# df.groupby("Team")["Salary"].std()
df.groupby("Department")["Salary"].std()
```

```python
# df.groupby("Team")["Salary"].agg(["mean", "max", "min"]).reset_index().rename_axis(None,axis=1)

df.groupby("Team")["Salary"].agg(["min","max","mean"]).reset_index().rename_axis(None,axis=1)
```

```python
# df.groupby("Team")["Salary"].aggregate(
#     avg_score="mean",
#     high_score="max"
# ).reset_index().rename_axis(None,axis=1)

df.groupby("Team")["Salary"].agg(avg_score="mean",high_score ="max").reset_index().rename_axis(None,axis = 1)
```
