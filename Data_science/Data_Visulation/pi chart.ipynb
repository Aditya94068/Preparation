```python
plt.pie(runs,
        labels = players,
        colors = colors,
        wedgeprops = {'edgecolor':'black','linewidth':1,'linestyle':'-','fill':True},
       shadow = True,
       startangle = 120)
```

```python
plt.pie(
    runs,
    labels=labels,
    autopct='%1.4f%%',
    startangle=140
)
plt.title("Career Run Share")
plt.show()
```
