```python
import numpy as np
sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
x = np.arange(len(years))
width =0.25
plt.bar(x - width,sachin,width =width , color = 'orange',label = "Sachin")
plt.bar(x,sehwag,width =width , color = 'red',label = "Sehwag")
plt.bar(x + width,kohli,width =width , color = 'blue',label = "Kohli")
plt.title("Comparison")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.xticks(x,years)
```

```python
# Create a side-by-side bar chart comparing runs of Sachin, Kohli, and Sehwag over 5 selected years.
# Then create a horizontal bar chart showing total runs scored in their debut 5 years.

years = [2000,2001,2002,2003,2004]
Sachin = [1000,3000,1200,3500,800]
Sehwag = [800,900,1250,1000,700]
Kohli = [1000,1200,800,2500,3000]
x = np.arange(len(years))
width = 0.25
plt.bar(x - width , Sachin,width = width,color = "red",label = "SACHIN")
plt.bar(x , Sehwag,width = width,color = "Orange",label = "Sehwag")
plt.bar(x + width , Kohli,width = width,color = "Blue",label = "Kohli")
plt.xlabel("Years")
plt.ylabel("Runs")
plt.xticks(x,years)
plt.legend()
plt.title("Comparison ")
```

```python
players = ["Sachin","Kohli","Sehwag"]
Runs_Five_Years = [ 0+200+900+1400+1600,500+700+1100+1500+1800, 0+0+500+800+1100]
plt.barh(players,Runs_Five_Years,color ="green")
plt.xlabel("Total in 5 years")
plt.ylabel("Players")
plt.title("Total Five year Runs of this cricket")
plt.tight_layout()
```
